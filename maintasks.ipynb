```python
def get_min(input_list: list):
    assert isinstance(input_list, list), "Входной параметр не является списком"
    assert len(input_list) > 0, "Входной параметр является пустым списком"
    current_min = input_list[0]
    min_indexes = ""
    for i in input_list:
        assert isinstance(i, (float, int, complex)) and not isinstance(i, bool), "Входной параметр имеет не только числовые значения"
    for i in range(len(input_list)):
        if input_list[i] < current_min:
            current_min = i
            min_indexes = i
        elif input_list[i] == current_min:
            min_indexes += str(i)
    return min_indexes
example_list = [5,-4,4556,123,44,-5,0,324]
min_value = get_min(example_list)
print(min_value)
```

```python
def get_bubble_sort_list(input_list: list):
    assert isinstance(input_list, list), "Входной параметр не является списком"
    assert len(input_list) > 0, "Входной параметр является пустым списком"
    for k in input_list:
        for i in input_list:
            assert isinstance(i, (float, int, complex)) and not isinstance(i, bool), "Входной параметр имеет не только числовые значения"
        for j in range(len(input_list)-1):
            if input_list[j] > input_list[j+1]:
                input_list[j], input_list[j+1] = input_list[j+1], input_list[j]
    return input_list
example_list = [9, 7, 3, 6, 2, 1, 8]
bubble_sort_list = get_bubble_sort_list(example_list)
print(bubble_sort_list)
```

```python
import random
def random_subsequence(input_number: int):
    letters = ["A", "G", "T", "C"]
    s = ""
    for i in range(input_number):
        s += random.choice(letters)
    return s
number = 10
subsequence = random_subsequence(number)
print(subsequence)
```

```python
import random
def guess_number(n):
    chosen_number = random.randint(1, 100)
    user_number = None
    attempts = 10
    while user_number != chosen_number and attempts > 0:
        user_number = int(input("Угадай число от 1 до 100"))
        if user_number < chosen_number:
            print("Загаданное число больше введенного")
            attempts -= 1
        elif user_number > chosen_number:
            print("Загаданное число меньше введенного")
            attempts -= 1
    if user_number == chosen_number:
        return "Поздравляю! Вы отгадали число!"
    else:
        return f"Попытки закончились. Загаданное число: {chosen_number}"
print(guess_number(100))
```
