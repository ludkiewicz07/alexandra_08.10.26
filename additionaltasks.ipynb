```python
def fibonacci(number):
    x = 0
    y = 1
    for i in range(number):
        x, y = y, x+y
    return x
example_number = 6
fib = fibonacci(example_number)
print(fib)
```

```python
def count_swaps(m):
    swaps = 0
    for i in range(1, len(m)):
        j = i
        while j > 0 and m[j] < m[j-1]:
            m[j], m[j-1] = m[j-1], m[j]
            swaps += 1
            j -= 1
    return swaps
massive = [6, 10, 4, 5, 1, 2]
mas = count_swaps(massive)
print(mas)
```

```python
def count_ACGT(n):
    counter_A = 0
    counter_C = 0
    counter_G = 0
    counter_T = 0
    final_count = ""
    for i in n:
        if i == "A":
            counter_A += 1
        elif i == "C":
            counter_C += 1
        elif i == "G":
            counter_G += 1
        elif i == "T":
            counter_T += 1
    final_count += str(counter_A) + " "+ str(counter_C) + " " + str(counter_G) + " " + str(counter_T)
    return final_count
DNA = "AGCTTTTCATTCTGACTGCAACGGGCAATATGTCTCTGTGTGGATTAAAAAAAGAGTGTCTGATAGCAGC"
dna = count_ACGT(DNA)
print(dna)
                
                
                
```

```python
def DNA_to_RNA(s):
    s = s.replace("T", "U")
    return s
DNA = "GATGGAACTTGACTACGTAAATT"
RNA = DNA_to_RNA(DNA)
print(RNA)
```

```python
def complementary_DNA(s):
    pairs = {"A": "T", "G": "C", "T": "A", "C": "G"}
    result = ""
    for i in range(-1, -len(s)-1, -1):
        result += pairs[s[i]]
    return result
DNA = "AAAACCCGGT"
dna = complementary_DNA(DNA)
print(dna)
```

```python
def count_rabbits(n, k):
    adult = 0
    young = 1
    for i in range(1, n):
        young, adult = adult * k, young + adult
    return adult + young
n = 5
k = 3
rab = count_rabbits(n, k)
print(rab)
```
